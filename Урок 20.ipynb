```python
# Создание словаря
user = {
    "id": 101,
    "name": "Alice",
    "is_active": True,
    "roles": ["admin", "user"]
}

# Доступ по ключу
print(user["name"])  # ✅ Alice
```

```python
# ❌ ОШИБКА: Если ключа нет, программа упадет
print(user["email"])  # KeyError: 'email'
```

```python
# ✅ БЕЗОПАСНО: Используем .get()
print(user.get("email", "not specified"))  # Вернет "not specified", если ключа нет
```

```python
text = "apple banana apple orange banana apple"
words = text.split()

counts = {}
for word in words:
    # Если ключа нет, создаем его со значением 0, затем прибавляем 1
    counts[word] = counts.get(word, 0) + 1

print(counts) 
# {'apple': 3, 'banana': 2, 'orange': 1}
```

```python
for word, cnt in counts.items():
    print(f'{word} встречается {cnt} раз')
```

```python
numbers = [1, 2, 3, 4, 5]

# Создать словарь {число: квадрат_числа}
squares = {x: x**2 for x in numbers}
print(squares)  # {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

# Фильтрация при создании (только четные)
even_squares = {x: x**2 for x in numbers if x % 2 == 0}
print(even_squares)  # {2: 4, 4: 16}
```
