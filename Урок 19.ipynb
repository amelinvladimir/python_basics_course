```python
# Получаем множество уникальных букв в слове

a = set('abracadabra')
b = set('alacazam')
a                                  # Уникальные буквы в a
```

```python
b
```

```python
a - b                              # Буквы, которые есть в a, но отсутствуют в b
```

```python
a | b                              # Буквы, которые есть или в a или в b
```

```python
a & b                              # Буквы, которые есть и в a и в b
```

```python
a ^ b                              # Буквы, которые есть или только в a или только в b
```

```python
# Роли пользователя
user_roles = {"editor", "viewer", "commenter"}

# Требования для удаления статьи
delete_permission = {"admin", "editor"}

# Требования для просмотра статьи
view_permission = {"viewer", "guest"}

# Проверка: Есть ли пересечение между ролями пользователя и требованиями?
if user_roles & delete_permission:  # Оператор & это intersection (пересечение)
    print("✅ Доступ к удалению разрешен")
else:
    print("❌ Недостаточно прав")

if user_roles & view_permission:
    print("✅ Доступ к просмотру разрешен")
```

```python
import time

# Генерируем большой список
big_list = list(range(10000000))
big_set = set(big_list)

# Проверка наличия числа 99999
start = time.time()
9999999 in big_list
print(f"List time: {time.time() - start}")  # Например: 0.002 сек (может быть больше на больших данных)

start = time.time()
9999999 in big_set
print(f"Set time: {time.time() - start}")   # Например: 0.00001 сек

# На больших данных разница будет в сотни раз!
```
