```python
attempts = 0

while True:
    try:
        s = input("Введите число: ")
        i = int(s)
        print(10 / i)
        break
    except ValueError: 
        print("Вы ввели не число")
    except ZeroDivisionError: 
        print("На ноль делить нельзя")
    finally:
        attempts += 1

print(f'Всего {attempts} попыток')
```

```python
attempts = 0

while True:
    try:
        s = input("Введите число: ")
        i = int(s)
        print(10 / i)
    except ValueError: 
        print("Вы ввели не число")
    except ZeroDivisionError: 
        print("На ноль делить нельзя")
    else:
        break
    finally:
        attempts += 1

print(f'Всего {attempts} попыток')
```

```python
def bool_return():
    try:
        return True
    finally:
        return False

bool_return()
```

```python
def get_int():
    s = input("Введите нечетное число: ")
    i = int(s)
    if i % 2 == 0:
        raise ValueError("Число делится на 2")
    return s

get_int()
```
