```python
# Ловушка новичка: кортеж из одного элемента
single_tuple = (1)      # ❌ Это просто число int!
print(single_tuple)

single_tuple = (1,)     # ✅ Это кортеж! Запятая обязательна.
print(single_tuple)

single_tuple = 1,       # ✅ Это кортеж! Без скобок.
print(single_tuple)
```

```python
def get_stats(numbers):
    return min(numbers), max(numbers), sum(numbers) / len(numbers)

# Распаковка сразу в переменные
minimum, maximum, average = get_stats([1, 2, 3, 4, 5])
print(minimum, maximum, average)
```
